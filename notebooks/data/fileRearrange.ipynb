```python
import shutil 
import os 

rootDir = 'A:\Lip Reading\Potential Datasets\BBC LRS2\mvlrs_v1\main'
newDir  = 'A:\Lip Reading\Potential Datasets\BBC LRS2\\allFiles'
newDir2  = 'A:\\Lip Reading\\Potential Datasets\\BBC LRS2\\allFiles2'
```

```python
def copyFileToLoc(file, subdir, dir, shouldPrint=False):
    fileFolder = subdir.split('\\')[-1]
    newFileName = fileFolder + '_' + file

    shutil.copy2(os.path.join(subdir, file), os.path.join(dir, newFileName))
    
    if shouldPrint: print(os.path.join(subdir, file))
```

```python
for subdir, dirs, files in os.walk(rootDir):
    for file in files:
        copyFileToLoc(file, subdir, newDir)
```

```python
# put one file from each folder into 
for subdir, dirs, files in os.walk(rootDir):
    for i in range(len(files)):
        # copyFileToLoc(file=files[0], subdir=subdir, dir=newDir2)
        txtFile = (files[0].split('.')[0] + '.txt')
        copyFileToLoc(file=txtFile, subdir=subdir, dir=newDir2, shouldPrint=True)
```
